```python
# 필요한 라이브러리 import
import sys
import os
from datetime import datetime
import json

# 프로젝트 루트를 Python path에 추가
project_root = os.path.abspath(os.path.join(os.getcwd()))
sys.path.insert(0, os.path.join(project_root, 'src'))

# 파이프라인 모듈 import
import run_qwen_exaone_pipeline as pipeline
import argparse

print("✓ Import 완료")
print(f"프로젝트 루트: {project_root}")
```

## 1. 파라미터 설정

아래 파라미터를 수정하고 실행하세요:

```python
# ========== 파라미터 설정 (여기를 수정하세요) ==========

PERSONA = 0  # 페르소나 인덱스 (0~9) 또는 이름
BRAND = "설화수"  # 브랜드명
PRODUCT = "자음생크림 리치 단품세트"  # 제품명
STAGE_INDEX = 1  # CRM 스테이지 (0=할인, 1=신제품런칭, 2=베스트셀러)
VIBE = "2030"  # 타겟 연령대 ("2030" 또는 "4060")
STYLE_INDEX = 0  # CRM 템플릿 스타일 (0~5)
IS_EVENT = 0  # 캠페인 이벤트 포함 여부 (0 또는 1)
TOP_K = 1  # RAG Top-K

# 모델 설정
QWEN_MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
EXAONE_MODEL = "LGAI-EXAONE/EXAONE-4.0-1.2B"

# ======================================================

print("파라미터 설정 완료:")
print(f"  - Persona: {PERSONA}")
print(f"  - Brand: {BRAND}")
print(f"  - Product: {PRODUCT}")
print(f"  - Stage: {STAGE_INDEX}")
print(f"  - Vibe: {VIBE}")
print(f"  - Style: {STYLE_INDEX}")
print(f"  - Event: {IS_EVENT}")
```

```python
# Arguments 생성
args = argparse.Namespace(
    persona=PERSONA,
    brand=BRAND,
    product=PRODUCT,
    stage_index=STAGE_INDEX,
    vibe=VIBE,
    style_index=STYLE_INDEX,
    is_event=IS_EVENT,
    top_k=TOP_K,
    qwen_model=QWEN_MODEL,
    exa_model=EXAONE_MODEL,
    disable_cache=False
)

print("="*80)
print("파이프라인 실행 중...")
print("="*80)

# 파이프라인 실행
result = pipeline._run_pipeline(args, data=data)

print("\n" + "="*80)
print("✓ 파이프라인 실행 완료")
print("="*80)
```

## 4. 결과 저장

실행 결과를 파일에 저장합니다.

```python
def save_results_to_file(result, test_dir="test"):
    """결과를 파일에 저장 (가독성 있게 포맷팅)"""
    os.makedirs(test_dir, exist_ok=True)
    
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    separator = "=" * 100
    sub_separator = "-" * 100
    
    # Qwen 결과 저장
    qwen_file = os.path.join(test_dir, "qwen_results.txt")
    with open(qwen_file, "a", encoding="utf-8") as f:
        f.write(f"\n\n{separator}\n")
        f.write(f"[Qwen 실행 결과] {timestamp}\n")
        f.write(f"{separator}\n\n")
        
        # 메타 정보
        f.write("[메타 정보]\n")
        f.write(f"  Persona: {result['persona_profile'].get('name', 'N/A')}\n")
        f.write(f"  Brand: {result['brand']}\n")
        f.write(f"  Product: {result['product_basic'].get('name', 'N/A')}\n")
        f.write(f"  Stage: {result['stage_name']} ({result['stage_kr']})\n")
        f.write(f"  Vibe: {result['vibe']}\n")
        f.write(f"  Context: {result['selected_context_example']}\n")
        f.write(f"\n{sub_separator}\n\n")
        
        # Qwen 입력 (Messages)
        f.write("[Qwen 입력 Messages]\n\n")
        qwen_messages = result['qwen'].get('draft', '')
        # 실제로는 messages를 출력하려면 파이프라인에서 저장해야 함
        # 여기서는 프롬프트 텍스트를 재구성
        f.write(f"Persona: {result['persona_profile'].get('name')}\n\n")
        
        behavioral_traits = result['persona_profile'].get('behavioral_traits', {})
        f.write("행동 특성:\n")
        for key, value in behavioral_traits.items():
            f.write(f"  - {key}: {value}\n")
        
        f.write(f"\n페인 포인트: {result['persona_profile'].get('pain_point', 'N/A')}\n")
        f.write(f"상황 컨텍스트: {result['selected_context_example']}\n")
        
        don_ts = result['persona_profile'].get('don_ts', [])
        if don_ts:
            f.write("\n금지 사항:\n")
            for item in don_ts:
                f.write(f"  - {item}\n")
        
        f.write(f"\n{sub_separator}\n\n")
        
        # Qwen 출력
        f.write("[Qwen 출력]\n\n")
        qwen_output = result['qwen']['draft']
        f.write(qwen_output)
        f.write("\n\n")
        
        # 타이밍 정보
        timing = result.get('timing', {})
        f.write(f"\n실행 시간: {timing.get('qwen', 0):.2f}초\n")
    
    # Exaone 결과 저장
    exaone_file = os.path.join(test_dir, "exaone_results.txt")
    with open(exaone_file, "a", encoding="utf-8") as f:
        f.write(f"\n\n{separator}\n")
        f.write(f"[Exaone 실행 결과] {timestamp}\n")
        f.write(f"{separator}\n\n")
        
        # 메타 정보
        f.write("[메타 정보]\n")
        f.write(f"  Persona: {result['persona_profile'].get('name', 'N/A')}\n")
        f.write(f"  Brand: {result['brand']}\n")
        f.write(f"  Product: {result['product_basic'].get('name', 'N/A')}\n")
        f.write(f"  Stage: {result['stage_name']} ({result['stage_kr']})\n")
        f.write(f"  Vibe: {result['vibe']}\n")
        f.write(f"  Context: {result['selected_context_example']}\n")
        f.write(f"\n{sub_separator}\n\n")
        
        # Exaone 입력 (Prompt)
        f.write("[Exaone 입력 Prompt]\n\n")
        exaone_prompt = result['exaone']['prompt_text']
        f.write(exaone_prompt)
        f.write("\n\n")
        f.write(f"{sub_separator}\n\n")
        
        # Persona 정보 (Exaone용)
        f.write("[Persona 정보 (Exaone)]\n")
        f.write(f"  이름: {result['persona_profile'].get('name')}\n")
        f.write(f"  상황 컨텍스트: {result['selected_context_example']}\n")
        f.write(f"  Vibe 톤 가이드: {result['vibe_text']}\n")
        f.write(f"\n{sub_separator}\n\n")
        
        # Exaone 출력
        f.write("[Exaone 출력]\n\n")
        exaone_output = result['exaone']['result_raw']
        f.write(exaone_output)
        f.write("\n\n")
        
        # 타이밍 정보
        f.write(f"\n실행 시간: {timing.get('exaone', 0):.2f}초\n")
    
    return qwen_file, exaone_file


# 결과 저장
qwen_file, exaone_file = save_results_to_file(result)

print("✓ 결과 저장 완료:")
print(f"  - Qwen 결과: {qwen_file}")
print(f"  - Exaone 결과: {exaone_file}")
```
